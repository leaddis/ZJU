这里介绍一个重要的`pandas`函数`merge()`，语法格式如下：
`merge ( left, right, how=' inner', on=None, left_on=None, right_on=None,left_index=False, right_index=False, sort=True, suffixes=('_x', 'y'), copy=True, indicator=False)`

参数介绍:
- left: 参与合并的左侧DataFrame;
- right: 参与合并的右侧DataFrame;
- how: 连接方式，有inner、left、right、outer，默认为inner；
- on: 指的是用于连接的列索引名称，必须存在于左右两个DataFrame中，如果没有指定且其他参数也没有指定，则以两个DataFrame列名交集作为连接键;
- left_on: 左侧DataFrame中用于连接键的列名，这个参数左右列名不同但代表的含义相同时非 常的有用;
- right_on: 右侧DataFrame中用于连接键的列名;
- left_index: 使用左侧DataFrame中的行索引作为连接键;
- right_index: 使用右侧DataFrame中的行索引作为连接键；
- sort: 默认为True，将合并的数据进行排序，设置为False可以提高性能;
- suffixes: 字符串值组成的元组，用于指定当左右DataFrame存在相同列名时在列名后面附加 的后缀名称，默认为('_x', '_y');
- copy: 默认为True，总是将数据复制到数据结构中，设置为False可以提高性能；
- indicator: 显示合并数据中数据的来源情况。

`inner`的类型称为内连接，它在拼接的过程中会取两张表的键（`key`）的交集进行拼接。

`left` 和 `right` 的 merge 方式其实是类似的，分别被称为左连接和右连接。
- `left`: merge 时，以左边表格的键为基准进行配对，如果左边表格中的键在右边不存在，则用缺失值 `NaN` 填充。
- `right`: merge 时，以右边表格的键为基准进行配对，如果右边表格中的键在左边不存在，则用缺失值 `NaN` 填充。

```python
df1 = pd.DataFrame({'userid':['a','b','c','d'], 'age':[23,46,32,19]})
df2 = pd.DataFrame({'userid':['a','c','e'], 'payment':[2000,3500,600]})
```

## 因子的分布密度

这里主要考察因子极端值对因子的影响，对比在去除极端值之后，因子分布密度的变化。去除极端值的方法主要有三种：
- 百分位取极值法
- 标准化去极值法
- 绝对值中位数法

### 百分位去极值：

是指直接以上下百分位为边界，将边界外数据归为边界上数据：

```python
# 百分位去除极端值
def winsor(x, lower = 5):
    upper = 100 - lower
    if x.dropna().shape[0] != 0:
        x.loc[x < np.percentile(x.dropna(),lower)] = np.percentile(x.dropna(),lower)
        x.loc[x > np.percentile(x.dropna(),upper)] = np.percentile(x.dropna(),upper)
    else:
        x = x.fillna(0)
    return x
```

### 标准化去极值法

标准化去极值方法又称为3σ去极值法。标准差$\sigma$本身可以体现因子的离散程度，是基于因子$X$的平均值$\mu_{X}$而定的。在离群值处理过程中，可通过用$[\mu_{X}-k\sigma, \mu_{X}+k\sigma]$来衡量因子与平均值的距离，其中$k=3$。$3 \sigma$ 法源于最经典的统计学 $3 \sigma$ 原则，即正态分布的数分布在 $(\mu-3 \sigma, \mu+3 \sigma)$ 中的概率为 $99.73 \%$, 在 $3 \sigma$ 外的概率是 $0.27 \%$ ，也就是图中尾部的那些数。$3 \sigma$ 去极值法其实就是把离平均值太远的值算作极端值，距离超过3倍标准差以上的就被认定为离均值太远。

```python
def winsorize_std(data,scale=3): 
    upper = data.mean() + scale*data.std()
    lower = data.mean() - scale*data.std()
    return np.clip(data,lower,upper)
```

### 绝对值中位数法

绝对值差中位数法(mean absolute deviation, MAD)通过计算所有因子与中位数之间的距离总和来检测离群值:
- 第一步：计算因子序列的中位数 $X_{median}$
- 第二步：计算每个值$X_{i}$与中位数的绝对偏差值 $|X_{i}-X_{median}|$
- 第三步：得到绝对偏差值的中位数MAD
- 第四步：确定参数 $k$, 从而确定合理的范围为 $\left[X_{median}-k \times \text{MAD}, X_{median}+k \times \text{MAD}\right]$
- 第五步：对超出合理范围的因子值做如下的调整超出最大值的用最大值代替，小于最小值的用最小值代替。

```python
def winsorize_MAD(data,scale=3): 
    median = data.quantile(0.5)
    MAD = ((data - median).abs()).quantile(0.50)
    upper = median + scale*MAD
    lower = median - scale*MAD
    return np.clip(data,lower,upper)
```

```python
#按年份画因子值的密度图
def plotFactor(data,if_winsor = 1):
    # data = allfactors.copy();savepath = output_picture
    """
    画因子分布图，hist，先做winsor
    """    
    fname = data.columns[2]
    
    data = f.copy()
    data['year'] = data.tradedate.apply(lambda x:x.year)
    data = data.fillna(0)
    
    if if_winsor == 1:            
        data = data.set_index(['tradedate','stockcode'])
        data = data.groupby('year').apply(lambda x:winsor(x[fname]) )   
        data = data.reset_index()
    elif if_winsor == 2:
        data = data.set_index(['tradedate','stockcode'])
        data = data.groupby('year').apply(lambda x:winsorize_MAD(x[fname]) )   
        data = data.reset_index()
    elif if_winsor == 3:
        data = data.set_index(['tradedate','stockcode'])
        data = data.groupby('year').apply(lambda x:winsorize_std(x[fname]) )   
        data = data.reset_index()
    else:
        pass
    
    allyear = data.year.unique()
    n = len(allyear)
    ncol = int(n/3)+1

    fig,axes = plt.subplots(3, ncol, figsize=(10, 7), sharex=True, sharey=True)

    for i in range(n):

        sns.kdeplot(data.loc[data.year ==allyear[i],fname],shade=True,ax=axes[int(i/ncol),i%ncol],label = allyear[i])

    plt.suptitle('Factor Distribution :' + fname)
```

```python
#  因子分布
plotFactor(f,0) # 不去除极端值
plotFactor(f,1) # 使用百分位法去极端值
```

```python
plotFactor(f,2) # 使用绝对值中位数方法去极端值
plotFactor(f,3) # 使用标准化法去极端值
```

可以看出，去除极端值前后，因子分布密度变化较大，说明极端值对后续分析会造成较大的影响，因此去除极端值的操作是有必要的。

## 因子的自相关分析

这里主要分析因子各阶自相关系数的变化，一般认为如果因子自相关系数下降过快，可能会导致依据该因子构建的投资组合股票换手率过高，增加交易费用，减少了组合收益。

```python
# 因子的自相关性分析
def getFactorADF(factors, lag = 12):
    # factors = fnormall;ret = retdata    
    """
    计算因子自相关性, factors是一个dataframe, 包含三列， 分别为stockcode, tradedate和mom。
    """
    factors = factors.copy()    
    factors = factors.set_index(['tradedate','stockcode'])
    fadf = []
    for i in factors.columns:       
        s = factors.loc[:,i].unstack().T.corr()
        lag_coe = []
        for j in range(lag):
            #fadf.append(pd.DataFrame(np.diag(s,j),columns = [i],index = s.index[1:]))
            lag_coe.append(np.diag(s,j).mean())
        fadf.append(pd.DataFrame(lag_coe, columns = [i] ))
    fadf = pd.concat(fadf,axis = 1)
    return fadf

def plotADF(fadf):  
        fig = plt.figure(figsize = (10,5))
        fadf.plot(kind = 'bar')
        #fadf.plot(kind = 'line',color = 'darkred',linewidth = 2)
        plt.title('因子自相关系数 :{}'.format(fadf.columns[0]),fontsize = 20)
```

```python
def plotIndf(f):
    '''
    f包括4列，列名分别为stockcode,tardedate,mom,classname
    '''
    fmean = f.groupby('classname').mean()
    fstd = f.groupby('classname').std()
    
    fname = fmean.columns[-1]        
    fnorm = pd.Series(fmean[fname]/fstd[fname]) 
    
    # 按照f大小进行排序
    fmean = fmean[fname].sort_values()
    fnorm = fnorm.loc[fmean.index]    
     
    plt.figure(figsize = (8,8)) 
    plt.barh(y = np.arange(fmean.shape[0]),height = 0.5,tick_label  = fmean.index,width = fmean,label = 'fmean')
    plt.barh(y = np.arange(fnorm.shape[0]) + 0.3,height = 0.5,tick_label = fnorm.index,\
             width = fnorm,label = 'fmean/fstd' )
    plt.title('因子行业分析 - ' + fname,fontsize = 20)
    plt.legend(fontsize = 20)
```

## 因子中性化

### 因子的行业中性化

### 因子市值中性化

```python
def OlsResid(y,x):
    df = pd.concat([y,x],axis = 1)

#    print(df)
    if df.dropna().shape[0]>0:    
        resid = sm.OLS(y,x,missing='drop').fit().resid   
        return resid.reindex(df.index)
    else:
        return y

def norm(data,if_neutral):
    """
    数据预处理，标准化
    """    
    # 判断有无缺失值，若有缺失值，drop，若都缺失，返回原值
    datax = data.copy()
    if data.shape[0] != 0:
        classname = data['classname']
        mkt = data['mktcap']    
        data = data.drop(['classname','mktcap'],axis = 1)
 
        ## 去极值
        data = data.apply(lambda x:winsor(x),axis = 0)
        ## 中性化
        if if_neutral:  # 是否中性
          
            class_var = pd.get_dummies(classname,columns=['classname'],prefix='classname',
                                       prefix_sep="_", dummy_na=False, drop_first=True)   
            class_var['mktcap'] = np.log(mkt)
            class_var['Intercept'] = 1
            x = class_var                
        # 每个因子对所有自变量做回归，得到残差值
            data = data.apply(func = OlsResid, args = (x,), axis = 0)                     
         
           ## zscore
        data1 = (data - data.mean())/data.std() 

        # 缺失部分补进去
        data1 = data1.reindex(datax.index)
    else:
        data1 = data                  
    return data1


"""
调用norm在对每月的因子做截面上的中性化
"""
def NormFactors(datas,if_neutral):
    # datas = factorall.copy()
    fnormall = []

    dates = datas.tradedate.unique()

    for dateuse in dates: # dateuse = dates[0]
        datause = datas.loc[datas.tradedate == dateuse]
        stockname = datause[['tradedate','stockcode']]
        fnorm = norm(datause.drop(['tradedate','stockcode'],axis = 1) ,if_neutral)
        fnormall.append(pd.concat([stockname,fnorm],axis = 1))
        print('{}中性化完成！'.format(dateuse))
    fnormall = pd.concat(fnormall,axis = 0)
    fnormall = fnormall.sort_values(by = ['tradedate','stockcode'])
    return fnormall.reset_index(drop = True)
```

### IC分析

对于两个随机变量 $X, Y$, 它们的皮尔逊相关系数定义为
$$
\rho=\operatorname{corr}(X, Y)=\frac{\operatorname{cov}(X, Y)}{\sqrt{\operatorname{var}(X) \cdot \operatorname{var}(Y)}} .
$$
设 $\left\{X_{i}, Y_{i}\right\}_{i=1}^{N}$ 是从总体 $\{X, Y\}$ 中抽取的样本, 则 $\rho$ 可被估计为
$$
\hat{\rho}=\frac{s_{x y}}{\sqrt{s_{x x} \cdot s_{y y}}}
$$
这里
$$
\begin{aligned}
s_{x y} &=\sum_{i=1}^{N}\left(X_{i}-\bar{X}\right)\left(Y_{i}-\bar{Y}\right) \\
s_{x x} &=\sum_{i=1}^{N}\left(X_{i}-\bar{X}\right)^{2} \\
s_{y y} &=\sum_{i=1}^{N}\left(Y_{i}-\bar{Y}\right)^{2}
\end{aligned}
$$
现在说明 $\hat{\rho}^{2}$ 等于一元线性回归模型中的可决系数。可决系数定义为
$$
R^{2}=\frac{\sum_{i=1}^{N}\left(\hat{Y}_{i}-\bar{Y}\right)^{2}}{\sum_{i=1}^{N}\left(Y_{i}-\bar{Y}\right)^{2}}
$$
注意到
$$
\begin{aligned}
&\hat{Y}_{i}=X_{i} \hat{\beta}_{1}+\hat{\beta}_{0}=X_{i} \hat{\beta}_{1}+\left(\bar{Y}-\bar{X} \hat{\beta}_{1}\right) \\
&\hat{\beta}_{1}=s_{x y} / s_{x x}
\end{aligned}
$$
因此
$$
\begin{aligned}
&\sum_{i=1}^{N}\left(\hat{Y}_{i}-\bar{Y}\right)^{2}=\sum_{i=1}^{N}\left(\hat{X}_{i}-\bar{X}\right)^{2} \hat{\beta}_{1}^{2} \\
&=s_{x x} \cdot \frac{s_{x y}^{2}}{s_{x x}^{2}}=s_{x y}^{2} / s_{x x} .
\end{aligned}
$$

从而得到
$$
R^{2}=\frac{\sum_{i=1}^{N}\left(\hat{Y}_{i}-\bar{Y}\right)^{2}}{\sum_{i=1}^{N}\left(Y_{i}-\bar{Y}\right)^{2}}=\frac{s_{x y}^{2}}{s_{x x} \cdot s_{y y}}=\hat{\rho}^{2} .
$$
接下来将说明检验相关系数是否为 0 的统计量等同于检验回リ系数是否为 0 的统计量。
现考虑假设检验问题
$$
H_{0}: \rho=0
$$
可构造如下统计量
$$
T_{N}=\frac{\sqrt{N-2} \hat{\rho}}{\sqrt{1-\hat{\rho}^{2}}}
$$
注意到以下结果
$$
\begin{aligned}
\frac{\sqrt{N-2} \hat{\rho}}{\sqrt{1-\hat{\rho}^{2}}} &=\frac{\hat{\rho}}{\sqrt{\left(1-\hat{\rho}^{2}\right) /(N-2)}} \\
&=\frac{s_{x y} / \sqrt{s_{x x} \cdot s_{y y}}}{\sqrt{\operatorname{SSE} /\left(s_{y y}(N-2)\right)}} \\
&=\frac{\hat{\beta}_{1} \sqrt{s_{x x}}}{\sqrt{\operatorname{SSE} /(N-2)}}=\text{t}_{N-2}.
\end{aligned}
$$
这 里 $\operatorname{SSE}=\sum_{i=1}^{n}\left(Y_{i}-\hat{Y}_{i}\right)^{2}$ 。第 二 个 等 号 是 由 $\hat{\rho}^{2}=R^{2}$ 以 及 $\mathrm{SSE}=\sum_{i=1}^{N}\left(Y_{i}-\bar{Y}\right)^{2}-\sum_{i=1}^{N}\left(\hat{Y}_{i}-\bar{Y}\right)^{2}$ 得到的。而最后的式子即为检验回归系数 $\beta_{1}$ 是否为 0 的 $\mathrm{t}$ 统计量, 其服从自由度为 $N-2$ 的 $\mathrm{t}$ 分布。

信息系数（Information Coefficient，简称IC）表示所选股票的因子值与股票下期收益率的截面相关系数，
$$
\text{IC} = \operatorname{corr}(F_{t}, R_{t+1})=\frac{\operatorname{cov}(F_{t}, R_{t+1})}{\sqrt{\operatorname{var}(F_{t}) \cdot \operatorname{var}(R_{t+1})}}
$$

通过 IC 值可以判断因子值对下期收益率的预测能力。通常将$t$期因子值和$t+1$期的收益率之间的信息系数，记作$\hat{\text{IC}}_{t,t+1}$：
$$
\hat{\text{IC}}_{t,t+1} = \frac{\frac{1}{T}\sum_{i=1}^{N}\left(f_{it}-\bar{f}_{t}\right)\left(r_{it+1}-\bar{r}_{t+1}\right)}{\sqrt{\frac{1}{T}\sum_{i=1}^{N}\left(f_{it}-\bar{f}_{t}\right)^2}\sqrt{\frac{1}{T}\sum_{i=1}^{N}\left(r_{it+1}-\bar{r}_{t+1}\right)^2}},
$$
这里我们设定：
$$
\bar{f}_{t} = \frac{1}{N}\sum_{i=1}^{N}f_{it},\quad \bar{r}_{t+1}=\frac{1}{N}\sum_{i=1}^{N}r_{it+1}.
$$
从上面的式子中我们可以计算得到一个信息系数的序列$\left\{\hat{\text{IC}}_{t,t+1}\right\}_{t=1}^{T-1}$.

可以得到
$$
(N-2)\frac{\hat{\text{IC}}_{t,t+1}^2}{1-\hat{\text{IC}}_{t,t+1}^2} = \text{t}_{N-2}^2,
$$
以及
$$
\hat{\text{IC}}_{t,t+1}^2 = R^{2}_{t,t+1}.
$$

```python
def getICSeries(factors,ret,method):
    # method = 'spearman';factors = fnorm;

    icall = pd.DataFrame()  
    fall = pd.merge(factors,ret,left_on = ['tradedate','stockcode'],right_on = ['tradedate','stockcode'])
    icall = fall.groupby('tradedate').apply(lambda x:x.corr(method = method)['ret']).reset_index()    
    icall = icall.dropna().drop(['ret'],axis = 1).set_index('tradedate')

    return icall    

def plotIC(ic_f):
    """
    IC作图
    """
    fname = ic_f.columns[0]
    
    fig = plt.figure(figsize = (9,4))
    ax = plt.axes()
    xtick = np.arange(0,ic_f.shape[0],18)
    xticklabel = pd.Series(ic_f.index[xtick])
    #每月因子的柱状图
    plt.bar(np.arange(ic_f.shape[0]),ic_f[fname],color = 'darkred')
    ax1 = plt.twinx()
    #因子按月累积值的图(画累计值有意义吗？请思考)
    ax1.plot(np.arange(ic_f.shape[0]),ic_f.cumsum(),color = 'orange')
    ax.set_xticks(xtick)
    ax.set_xticklabels(xticklabel)    
    plt.title(fname + '  IC = {},ICIR = {}'.format(round(ic_f[fname].mean(),4),\
                                                   round(ic_f[fname].mean()/ic_f[fname].std()*np.sqrt(12),4)))  
```

```python
#计算每个月的IC
icall = getICSeries(fnorm,ret,'spearman')
```

```python
def describeICSeries(factors,ret, icall):

    fname = icall.columns[0]
    
    fall = pd.merge(factors,ret,left_on = ['tradedate','stockcode'],right_on = ['tradedate','stockcode'])
    
    icall_n = pd.DataFrame(fall.groupby('tradedate').apply(lambda x:x.shape[0]).reset_index()).rename(columns={0:'n'})
    icall_test =  pd.merge(icall,icall_n,left_on = ['tradedate'],right_on = ['tradedate'])
    icall_test_statistic = pd.DataFrame(np.sqrt(icall_test.n-2)*icall_test[fname]/np.sqrt(1-icall_test[fname]**2)).set_index(icall_test.tradedate)
    icall_test_statistic = icall_test_statistic.rename(columns={0:'icall_test_statistic'}).reset_index()
    
    
    IC_mean = icall[fname].mean()
    IC_std  = icall[fname].std()
    if_IC_0 = np.mean(icall[fname]>0)
    if_IC_thereshhold = np.mean(abs(icall[fname])>0.03)
    # 适用于Pearson IC, 当样本量大于50，也适用于Spearman IC
    if_IC_test = np.mean(abs(icall_test_statistic['icall_test_statistic'])>1.96)
    
    ICIR    = IC_mean/IC_std*np.sqrt(12)
    IC_t    = np.sqrt(icall.shape[0])*ICIR
    IC_result = {"IC样本均值":round(IC_mean, 4),
              "IC样本标准差":round(IC_std, 4),
              "IC>0比例":round(if_IC_0, 4),
              "IC检验统计量>1.96比例":round(if_IC_test, 4),
              "abs(IC)>0.03比例":round(if_IC_thereshhold, 4),
              "年化ICIR":round(ICIR, 4),
              "IC-FM检验t值": round(IC_t, 4)
              }

    return IC_result
```

#### 分行业计算IC

```python
# 因子行业IC、ICIR作图

def getIndIC(factors,ret,classdata,method = 'spearman'):
    if 'classname' not in factors.columns:
        factors = pd.merge(factors,classdata,left_on = ['tradedate','stockcode'],right_on = ['tradedate','stockcode'])

    fall = pd.merge(factors,ret,left_on = ['tradedate','stockcode'],right_on = ['tradedate','stockcode'])
    findcorr = fall.groupby(['tradedate','classname']).apply(lambda x:x.corr(method = method)['ret']).reset_index()
    ic_ind = findcorr.drop(['ret'],axis = 1).dropna().reset_index(drop = True)


    return ic_ind

def plotIndIC(ic_ind):

    icind_mean = ic_ind.groupby('classname').mean()
    icir_ind = icind_mean/ic_ind.groupby('classname').std()*np.sqrt(12)
    
    
    for f in icind_mean.columns: # f = icind_mean.columns[2]
        # 按照IC大小进行排序
        icmean_use = icind_mean[f].sort_values()
        icir_ind_use = icir_ind.loc[icmean_use.index]    
        icuse = ic_ind[['tradedate','classname',f]].pivot(index = 'tradedate',columns = 'classname',values = f)
        icuse = icuse.loc[:,icmean_use.index]
    
        # IC、ICIR柱状图
        plt.figure(figsize = (8,8)) 
        plt.barh(y = np.arange(icmean_use.shape[0]),height = 0.5,tick_label  = icmean_use.index,width = icmean_use,label = 'IC')
        plt.barh(y = np.arange(icmean_use.shape[0]) +0.3,height = 0.5,tick_label = icir_ind_use.index,width = icir_ind_use[f],label = 'ICIR')
        plt.title('因子行业IC均值/行业ICIR - ' + f,fontsize = 15)
        plt.legend()

    
        # 累计IC
        icuse.cumsum().plot(figsize = (8,6),linewidth = 2,cmap = 'rainbow',legend = False)
        plt.legend(fontsize = 8,bbox_to_anchor = (1.05,1.05))
        plt.title('因子行业累计IC - ' + f,fontsize = 15)

    return icind_mean,icir_ind
```

#### 因子IC信息衰减

信息具有其时效性，对于选股因子而言，每个股票收益率与因子值的相关性可以看作投资者所观察到的“信息”，该信息同样具有时效性，一般情况下，随着预测时长的增加，信息量将逐步下降。因子IC的时间衰减是被提及最多的一个因子衰减概念，用以衡量一个因子对末来的预测能力能持续 多久。当一个指标时间衰减过快时，可能会导致组合较高的换手，交易成本会大幅侵蚀模型的盈利能力。

- 计算$t$期因子值和$t+1+\tau$期的收益率之间的IC的估计值，记作$\hat{\text{IC}}_{t,t+1+\tau}$：
$$
\hat{\text{IC}}_{t,t+1+\tau} = \frac{\frac{1}{T}\sum_{i=1}^{N}\left(f_{it}-\bar{f}_{t}\right)\left(r_{it+1+\tau}-\bar{r}_{t+1+\tau}\right)}{\sqrt{\frac{1}{T}\sum_{i=1}^{N}\left(f_{it}-\bar{f}_{t}\right)^2}\sqrt{\frac{1}{T}\sum_{i=1}^{N}\left(r_{it+1+\tau}-\bar{r}_{t+1+\tau}\right)^2}},
$$
这里我们设定：
$$
\bar{f}_{t} = \frac{1}{N}\sum_{i=1}^{N}f_{it},\quad \bar{r}_{t+1+\tau}=\frac{1}{N}\sum_{i=1}^{N}r_{it+1+\tau}.
$$

- 计算序列$\left\{\hat{\text{IC}}_{t,t+1+\tau}\right\}_{t=1}^{T-1-\tau}$的均值：
$$
\hat{\text{IC}}_{\tau}=\frac{1}{T}\sum_{t=1}^{T-1-\tau}\text{IC}_{t,t+1+\tau}.
$$

- 序列$\left\{\hat{\text{IC}}_{\tau}\right\}_{\tau=0}^{T-2}$的半衰期定义为$\text{H}$，满足条件：
$$
\text{H} = \text{argmin}_{\tau}\left\{\tau:\hat{\text{IC}}_{\tau} \leq \frac{1}{2}\hat{\text{IC}}_{0}\right\}. 
$$

可以简单的用IC半衰期来衡量IC时间衰减的快慢，由于我们主要是以月频来计算因子的IC，因此IC半衰期可定义为月度IC第一次下降到一半或一半以下所用的时间。同理，我们也可以计算 IC_IR的半衰期和衰减速度。

```python
def getHalfValue(factors,method,ret):
    """
    计算因子半衰期
    """
    # factors = fnormall;ret = retdata
    halfic = []     
    for i in range(12):
        forward = i+1
        ret1 = ret.pivot(index = 'tradedate',columns = 'stockcode',values = 'ret').shift(-forward).stack().reset_index()
        ret1 = ret1.rename(columns = {ret1.columns[-1]:'ret'})
        ic = getICSeries(factors,ret1,method)  
        print('滞后{}期IC完成！'.format(i))
        # 注意这里计算的是IC序列的均值
        halfic.append(pd.DataFrame(ic.mean(),columns = ['lag_' + str(i)]))
    halfic = pd.concat(halfic,axis = 1)

    halficvalue = halfic.apply(calhalfic,axis = 1)    
    return halfic,halficvalue

def plotHalfIC(halfic):
    halficvalue = halfic.apply(calhalfic,axis = 1)   
    for i in range(halfic.shape[0]): # i= 1
        fig = plt.figure(figsize = (10,5))
        halfic.iloc[i].plot(kind = 'bar')
        halfic.iloc[i].plot(kind = 'line',color = 'darkred',linewidth = 2)
        plt.title('IC衰减 :{},半衰期 = {}'.format(halfic.index[i],str(halficvalue[i]).strip('>')),fontsize = 20)

def calhalfic(x):
    """
    计算因子半衰期的函数
    """
    target = abs(x[0]/2)
    position = np.where(x.abs() < target)
    if len(position[0]) >0:
        return position[0][0]
    else:
        return '>{}'.format(len(x))
```

```python
# ic衰减
halfic,halficvalue = getHalfValue(fnorm,'spearman',ret)
```

#### 分组IC分析

在每一时期，将股票根据因子值的大小按照分位数分成$n_{p}$组，计算每组的IC，然后计算每组IC的时序平均值。

```python
def getGroupIC(fdata,rt,method,groups):
    # groups = 5
    """
    计算给定日期，因子分组IC
    
    groups为分组数
    
    fdata为包含给定日期因子值的dataframe，index为股票代码，
    column为因子值, 例如：这里列名为mom
    
    rt为包含给定日期收益率的dataframe，index为股票代码，
    column为收益率, dataframe的名字为给定的日期
    
    result为一个dataframe，包含一行，五列，列名为别为，0，1，2，3，4，
    分别代表给定的日期，第一组，第二组，第三组，第四组，第五组， index名字为startdate，包含
    因子值和收益率所属的日期。
    """
    # 使用pd.concat()之后，得到一个新的dataframe，包括两列
    # 第一列为因子值，列名mom，第二列为收益率，列名为日期， 比如：2009-12-31
    rt = pd.concat([fdata,rt],axis = 1).dropna()  
    indexs = ["startdate"] + list(range(groups))
    if rt.shape[0] != 0:
        # 根据因子值序列进行分组，得到给定日期每一只股票的分组标记
        groupdata = pd.qcut(rt.iloc[:,0],q = groups,labels = False,duplicates = 'drop')
        # 判断分组数目是否和预先设定的相同
        if groupdata.unique().shape[0] == groups:
            # 在rt中新加入一列，名字是group，取值为groupdata，标记出每一只股票的分组
            rt['group'] =  groupdata
            # 分组计算因子值和收益率的相关系数，x.corr()计算的是相关系数矩阵
            # .iloc[0,1]是取出矩阵的对角线元素，就是correlation coefficient
            IC = rt.groupby('group').apply(lambda x:x.corr(method = method).fillna(0).iloc[0,1])    
    
            result = pd.DataFrame([rt.columns[1]] + IC.tolist(),index = indexs).T   
        else:
            result = pd.DataFrame([rt.columns[1]] + [0]*groups,index = indexs).T             
    else:
        result = pd.DataFrame([rt.columns[1]] + [0]*groups,index = indexs).T           
    return result.set_index('startdate')
```

```python
def getGroupICSeries(factors,ret,method,groups):
    
    '''
    计算所有日期的因子分组IC值
    Parameters
    ----------
    factors : 数值型
        dataframe包括3列，列名分别为tradedate, stockcode和因子的名称，这里为mom.
        index为默认的index
    ret : 数值型
        dataframe包括3列，列名分别为tradedate, stockcode和ret，ret为股票收益率.
        index为默认的index
    method : object
        取值为”pearson“ 或者 ”spearman“.
    groups : TYPE
        可以取数字，5或者10，代表按照quantile分成5组或者10组.

    Returns
    -------
    icall : 数值型
        dataframe, 包括6列，列名分别为factor，0，1，2，3，4. index名为startdate,
        代表日期

    '''
    # method = 'spearman';factors = fnorm.copy();ret = ret
    icall = pd.DataFrame()  
    
    dates = factors.tradedate.unique()
    ret = ret.pivot(index = 'tradedate',columns = 'stockcode',values = 'ret')
    for dateuse in dates:  # dateuse = dates[0]      
        fic = pd.DataFrame()
        # 取出给定日期可以获得的所有股票的因子数据，将stockcode这一列设置为index         
        fdata = factors.loc[factors.tradedate == dateuse,factors.columns[1:]].set_index('stockcode') 
        # 取出给定日期可以获得的所有股票的收益率数据
        rt = ret.loc[dateuse] 
        '''
        # 这里使用循环的原因是可以应对因子多于一个的情况
        for f in fdata.columns: # f = fdata.columns[0] #如果只有一个因子，使用这条命令就可以
            IC = getGroupIC(fdata[f],rt,method,groups)
            IC.insert(0,'factor',f)
            fic = pd.concat([fic,IC],axis = 0)
        '''
        f = fdata.columns[0] #如果只有一个因子，使用这条命令就可以
        IC = getGroupIC(fdata[f],rt,method,groups)
        IC.insert(0,'factor',f)
        fic = pd.concat([fic,IC],axis = 0)
        
        icall = pd.concat([icall,fic],axis = 0) 

    return icall   

def plotGroupIC(groupIC):
    """
    分组IC作图，比较每组IC在所有月份的平均值
    """
    for f in groupIC.factor.unique():
        fig = plt.figure(figsize = (10,5))
        groupIC.loc[groupIC.factor == f,groupIC.columns[1:]].mean(axis =0).plot(kind = 'bar')
        plt.title('因子分组IC均值 : ' + f,fontsize = 20)
```

```python
# 因子分组IC

groups = 5
method = 'spearman'
groupIC = getGroupICSeries(fnorm,ret,method,groups)
     
```

#### `cut()`和`qcut()`函数的基本介绍

在pandas中，`cut()`和`qcut()`函数都可以进行分箱处理操作。其中`cut()`函数是按照数据的值进行分割， 而`qcut()`函数则是根据数据本身的数量来对数据进行分割。

-  `cut()`函数

```python
d = pd.DataFrame([x**2 for x in range(11)], columns=['number'])
print(d)
```

`cut()`函数是按变量的值进行分割

```python
# cut: 按照数据值由小到大的顺序将数据分成4份, 并且使每组值的范围大致相等。
d_cut = d.copy()
d_cut['cut_group']= pd.cut(d_cut['number'], 4)
d_cut
```

我们可以看到，上面的代码把数据按照由小到大的顺序平均切分成了4份，每份的值的跨度大约是25。其中， $(a1, a2]$ 表示 $a1<x\leq a2$ ，默认情况下，每个区间包括最大值，不包括最小值。但是最左边的值，一般设置成最小值 $(0)$ 减去最大值 $(100)$ 的 $0.1 \%$, 也就是 $0-100^{*} 0.1 \%=-0.1$ 。

```python
# 查看每个分组里变量的个数
d_cut['cut_group'].value_counts()
```

- `qcut()`函数

跟`cut()`函数按照变量的值对变量进行分割不同，`qcut()`函数按照分位数对变量进行分割。

```python
# 把变量由小到大分成四组，并且让每组变量的数量相同
d_qcut = d.copy()
d_qcut['qcut_group'] = pd.qcut(d_qcut['number'], 4)
d_qcut
```

- `cut()`和`qcut()`函数的拓展用法

### 分层组合分析

#### 因子的换手率分析

换手率通过每一期买入股票的变动，评价因子的交易成本暴露，实际计算时采用的是两期之间股票变动的数量除以上期持有的总的股票数量。换手率越高说明交易次数越多，可能产生的交易手续费越高。假设我们根据因子值的大小买入股票，那么因子值可以看作是一个评分，如果一个因子，第一期给一个股票打了很高的分，第二期就打很低的分，这将导致对该股频繁的买卖，从而造成很高的手续费。

```python
def getturnover(data,if_plot = False):
    # data = stocklist1.copy()
    
    turnover = pd.DataFrame()
    turnover['tradedate']= data.tradedate.unique()
    turnover['turnover'] = 0
    
    for i in range(1,turnover.shape[0]):
        stock_now = set(data.loc[data.tradedate==turnover.tradedate[i],'stockcode'].tolist())
        stock_pre = set(data.loc[data.tradedate==turnover.tradedate[i - 1],'stockcode'].tolist())
        turnover.loc[i,'turnover'] = len(stock_pre.difference(stock_now))/len(stock_pre)       
    
#    turnover['tradedate'] = turnover.tradedate.apply(lambda x:x.date().strftime("%Y%m%d"))
#    turnover = turnover.set_index('tradedate')
#    
    if if_plot:
        xtick = np.arange(0,turnover.shape[0],12)
        xticklabel = pd.Series(turnover.index[xtick])  
        
        plt.figure(figsize = (8,4))
        ax = plt.axes()
    
        plt.bar(np.arange(turnover.shape[0]),turnover.turnover,color = 'darkred',width = 0.2)       
        ax.set_xticks(xtick)
        ax.set_xticklabels(xticklabel) 
        plt.title('组合换手率',fontsize = 15)
 
    return turnover.set_index('tradedate')
```

#### 分组收益率分析

```python
def GroupTestAllFactors_1(factors,ret,groups):
    # factors = fnorm.copy();groups = 10
    '''
    计算所有日期的因子分组IC值
    Parameters
    ----------
    factors : 数值型
        dataframe包括3列，列名分别为tradedate, stockcode和因子的名称，这里为mom.
        index为默认的index
    ret : 数值型
        dataframe包括3列，列名分别为tradedate, stockcode和ret，ret为股票收益率.
        index为默认的index
    
    groups : TYPE
        可以取数字，5或者10，代表按照quantile分成5组或者10组.

    Returns
    -------
   

    '''
    fnames = factors.columns
    fall = pd.merge(factors,ret,left_on = ['stockcode','tradedate'],right_on = ['stockcode','tradedate'])
    Groupret = []
    Groupturnover = []
    
    f= fnames[2]
    if ((f != 'stockcode')&(f != 'tradedate')):
            fuse = fall.copy()[['stockcode','tradedate','ret',f]]
            # 计算每只股票在每个日期所属的group，按照因子值从小到大等分，注意这里
            # 每组的标志分别为1，2，3，4，5
            fuse['groups'] = fuse[f].groupby(fuse.tradedate).apply(lambda x:np.ceil(x.rank()/(len(x)/groups)))
            # 将每个日期每个group中的股票等权重构造投资组合，并计算组合收益率
            result = fuse.groupby(['tradedate','groups']).apply(lambda x:x.ret.mean())
            # 构建一个透视表，使用tradedate, groups和收益率
            result = result.unstack().reset_index()
            # 构建多空投资组合
            # 如果第五组的收益高于第一组，构建高减低投资组合，保存每个日期属于第五组的股票
            if result.iloc[:,-1].mean() > result.iloc[:,-groups].mean():
                result['L-S'] = result.iloc[:,-1] - result.iloc[:,-groups]
                #stock_l = fuse.loc[fuse.groups == 1]
                stock_l = fuse.loc[fuse.groups == groups]
            else: # 如果第一组的收益高于第五组，构建低减高投资组合，保存每个日期属于第一组的股票
                result['S-L'] = result.iloc[:,-groups] - result.iloc[:,-1]
                #stock_l = fuse.loc[fuse.groups == groups]
                stock_l = fuse.loc[fuse.groups == 1]
                
    result.insert(0,'factor',f)
    Groupret.append(result)
    
    turnovers = getturnover(stock_l)
    turnovers.columns = [f]
    # 计算换手率
    Groupturnover.append(turnovers)
    
    Groupret = pd.concat(Groupret,axis = 0).reset_index(drop = True)
    
    # 计算每一组的累积收益率
    Groupnav = Groupret.iloc[:,2:].groupby(Groupret.factor).apply(lambda x:(1 + x).cumprod())
    Groupnav = pd.concat([Groupret[['tradedate','factor']],Groupnav],axis = 1)
    # 计算换手率
    Groupturnover = pd.concat(Groupturnover,axis = 1).reset_index()

    return Groupret, Groupnav,Groupturnover
```

```python
def plotnav(Groupnav):
    """
    GroupTest作图
    """
    for f in Groupnav.factor.unique(): # f = Groupnav.factor.unique()[0]
        fnav = Groupnav.loc[Groupnav.factor ==f,:].set_index('tradedate').iloc[:,1:]
        groups = fnav.shape[1] - 1
        lwd = [2]*groups + [4]
        ls = ['-']*groups + ['--']
        
        plt.figure(figsize = (10,5))        
        for i in range(groups+1):
            plt.plot(fnav.iloc[:,i],linewidth = lwd[i],linestyle = ls[i])
        plt.legend(list(range(groups)) + ['L-S'])
        plt.title('因子分层测试/多空组合 ： ' + f,fontsize = 20)
```

- 最大回撤

最大回撤是指投资组合在选定的周期内，任一时间点往后推，可能出现资产净值下降的最大幅度。回撤的意思是指在某一段时期内净值从最高点开始回落到最低点的幅度。最大回撤常用百分率来表示，是一个重要的风险指标。最大回撤的计算公式为
$$
最大回撤 =\frac{波峰值 - 波谷值}{波峰值}
$$
注意这里的波峰值与波谷值并不一定是最高值与最低值，这里的波峰值与波谷值要与时间范围内关联起来。

如果考虑净值下降值，不采用百分率来表示下降幅度，即对于序列 $x_{1}, x_{2}, \ldots, x_{n}$，定义最大 回撤 $d$ 为
$$
d=\max _{i \leqslant j}\left(x_{i}-x_{j}\right)
$$
如果需要改用百分率来表示最大回撤，只需要再除以$x_{i}$即可。
根据定义，很容易想到最大回撤的算法。遍历序列中所有元素，对于每个元素计算其下降最大值，然后再比较所有元素下降最大值，挑选 出最大的元素下降最大值。这个算法的时间复杂度为 $O\left(n^{2}\right)$。
那么，是否存在线性时间复杂度求最大回撤的算法呢? 答案是肯定的。算法可以直接使用以下式子来表示:
$$
d=\max _{i \leqslant j}\left(x_{i}-x_{j}\right)=\max _{j}\left(\max _{i \leqslant j} x_{i}-x_{j}\right)
$$

这个式子看着复杂，其实不难理解。式子左边是最大回撤的定义。下面介绍如何理解式子的右半部分。我们先来看 $\left(\max _{i \leqslant j} x_{i}-x_{j}\right)$ ，假设我们指定元素 $x_{j}$ ，元素 $x_{i}$ 是指所有在元素 $x_{j}$ 前的元素。 $\max _{i \leqslant j} x_{i}$ 表示所有在元素 $x_{j}$ 前的元素中的最大元 素，$\left(\max _{i \leqslant j} x_{i}-x_{j}\right)$ 则表示这个最大元素减去元素 $x_{j 。}$ 说得或许还有点抽象，我们以一个具体例子来说明。
假设有一序列,$[100,200,50,300,150,100,200]$，元素下标以$0$来开始，指定 $j$ 为 $4，x_{j}$ 为 150，则可以取值 $0,1,2,3,4 ，$ 故 $\max _{i \leqslant j} x_{i}$ 为 $\left.300 （ \max _{i \leqslant j} x_{i}-x_{j}\right)$ 为 150 。
现在再来看 $\max _{j}\left(\max _{i \leqslant j} x_{i}-x_{j}\right)$ 来含义。前面我们可以计算到对于每个 $j$ ，都有一个 $\left(\max _{i \leqslant j} x_{i}-x_{j}\right)$ 值，对于所有的 $j$，取一个最大的 $\left(\max _{i \leqslant j} x_{i}-x_{j}\right)$值，即为最大回撤值，并取出取得最大回撤时对应的$j$和$i$下标，就完成算法的求解。

同理，最大回撤的百分率可以采用如下公式计算：
$$
\mathop{\arg\max}_{j}\left\{\max _{i \leqslant j}\left(\frac{x_{i}-x_{j}}{x_{i}}\right)\right\}=\mathop{\arg\max}_{j}\left\{\max _{i \leqslant j}\left(1-\frac{x_{j}}{x_{i}}\right)\right\} =\mathop{\arg\max}_{j}\left\{\min_{i \leqslant j}\left(\frac{x_{j}}{x_{i}}\right)\right\}=\mathop{\arg\max}_{j}\left\{\max_{i \leqslant j}\left(\frac{x_{i}}{x_{j}}\right)\right\}
$$
这里我们要求所有的$x_{i}$和$X_{j}$都为正值。

```python
# 计算最大回撤比率
def maxdrawdown_vec(arr):
    # arr是每期收益率的时间序列
    # 先计算累积收益率
    arr = (1+arr).cumprod()
    j = np.argmax((np.maximum.accumulate(arr) - arr)/np.maximum.accumulate(arr)) # end of the period
    i = np.argmax(arr[:j]) # start of period
    return (1-arr[j]/arr[i])

def maxdrawdown_vec_1(arr):
    # arr是每期收益率的时间序列
    # 先计算累积收益率
    arr = (1+arr).cumprod()
    j = np.argmax(np.maximum.accumulate(arr) /arr) # end of the period
    i = np.argmax(1 - arr[j]/arr[:j]) # start of period
    return (1-arr[j]/arr[i])
```

- Sharpe Ratio （夏普比率）

William Forsyth Sharpe is a Nobel-prize winning economist, who helped create the $\underline{\text { Capital Asset Pricing Model (CAPM)}}$  and developed the Sharpe Ratio in 1966 (later updated in 1994).
The Sharpe Ratio $S$ is defined by the following relation:
$$
S=\frac{\mathbb{E}\left(R_{a}-R_{b}\right)}{\sqrt{\operatorname{Var}\left(R_{a}-R_{b}\right)}}
$$
Where $R_{a}$ is the period return of the asset or strategy and $R_{b}$ is the period return of a suitable benchmark.
The ratio compares the mean average of the excess returns of the asset or strategy with the standard deviation of those returns. Thus a lower volatility of returns will lead to a greater Sharpe ratio, assuming identical returns.
The "Sharpe Ratio" often quoted by those carrying out trading strategies is the annualised Sharpe, the calculation of which depends upon the trading period of which the returns are measured. Assuming there are $N$ trading periods in a year, the annualised Sharpe is calculated as follows:
$$
S_{A}=\sqrt{N} \frac{\mathbb{E}\left(R_{a}-R_{b}\right)}{\sqrt{\operatorname{Var}\left(R_{a}-R_{b}\right)}}
$$
Note that the Sharpe ratio itself MUST be calculated based on the Sharpe of that particular time period type. For a strategy based on trading period of days, $N=252$ (as there are $252$ trading days in a year, not $365$ ), and $R_{a}$, $R_{b}$ must be the daily returns. Similarly for hours $N=252 \times 6.5=1638$, not $N=252 \times 24=6048$, since there are only $6.5$ hours in a trading day.

```python
# 计算年化夏普率
def annualised_sharpe_ratio(arr, periods = 12):
    """
    Create the annualised Sharpe Sharpe ratio for the strategy, based on a 
    benchmark of zero (i.e. no risk-free rate information).

    Parameters:
    returns - A pandas Series representing period percentage returns.
    periods - Monthly(12), Daily (252), Hourly (252*6.5), Minutely(252*6.5*60) etc.
    """
    sharperatio = np.sqrt(periods)*arr.mean()/arr.std()
    return sharperatio

# 计算年化平均收益率
def annualised_mean(arr, periods = 12):
    """
    Create the annualised mean return for the strategy, based on a 
    benchmark of zero (i.e. no risk-free rate information).

    Parameters:
    returns - A pandas Series representing period percentage returns.
    periods - Monthly(12), Daily (252), Hourly (252*6.5), Minutely(252*6.5*60) etc.
    """
    mean_return = np.power(1+arr.mean(), periods) - 1
    return mean_return

# 计算年化波动率
def annualised_std(arr, periods = 12):
    """
    Create the annualised volatility for the strategy, based on a 
    benchmark of zero (i.e. no risk-free rate information).

    Parameters:
    returns - A pandas Series representing period percentage returns.
    periods - Monthly(12), Daily (252), Hourly (252*6.5), Minutely(252*6.5*60) etc.
    """
    std = np.sqrt(periods)*arr.std()
    return std
```

```python
# 因子分层组合绩效分析
def groupAnalysis(Groupret):
    
    result = Groupret.iloc[:,2:].agg([annualised_mean, annualised_std, annualised_sharpe_ratio, maxdrawdown_vec])
    
    return result

def groupAnalysis_1(Groupret):
    
    result = Groupret.iloc[:,2:].agg([annualised_mean, annualised_std, annualised_sharpe_ratio, maxdrawdown_vec_1])
    
    return result
    
def groupAnalysis_excess_1(Groupret, benchmark):
    '''
    计算超额收益的分组统计量

    Parameters
    ----------
    Groupret : TYPE
        DESCRIPTION.
    benchmark : TYPE
        DESCRIPTION.

    Returns
    -------
    result_analysis : TYPE
        DESCRIPTION.

    '''
    result = Groupret.iloc[:,1:].set_index('tradedate').shift(1).dropna()
    #result = Groupret.iloc[:,2:].agg([annualised_mean, annualised_std, annualised_sharpe_ratio, maxdrawdown_vec_1])
    # benchmark是日度数据
    benchmark['ym'] = pd.Series(benchmark.tradedate).apply(lambda x:x.year*100 + x.month)
    # 计算benchmark的月度收益率
    # 保留每个月最后一个交易日的价格用来计算月度收益率
    benchmark = benchmark.groupby('ym').last()
    benchmark = benchmark.set_index('tradedate')[['price']].pct_change(1)
    benchmark = benchmark.reindex(result.index)
    
    result['benchmark'] = benchmark.price.values
    
    result_excess = result.apply(lambda x: x - result['benchmark']).drop(['benchmark'], axis=1)
    
    result_analysis = result_excess.agg([annualised_mean, annualised_std, annualised_sharpe_ratio, maxdrawdown_vec_1])
    
    return result_analysis    
```

#### Top N组合分析

```python
def portfolioTopN(factors,rt,benchmark,N,if_max = True):
    # rt = ret1.copy();benchmark = price300.copy();factors = fnormall.copy()

    if if_max:
        factors = factors.sort_values(by = factors.columns[2],ascending = 0)
        stocklist = factors.groupby('tradedate').head(N).reset_index(drop = True)
    else:
        factors = factors.sort_values(by = factors.columns[2],ascending = 1)                
        stocklist = factors.groupby('tradedate').head(N).reset_index()        
    stockret = pd.merge(stocklist,rt,left_on = ['tradedate','stockcode'],right_on = ['tradedate','stockcode'])
    stockret = stockret.sort_values(by = ['tradedate','stockcode'])
    
    result = stockret.ret.groupby(stockret.tradedate).mean()
    result  = pd.DataFrame(result).shift(1).fillna(0)
    result.columns = ['Portfolio']
    result['Portfolio'] = (result.Portfolio+1).cumprod()
    
    
    benchmark['ym'] = pd.Series(benchmark.tradedate).apply(lambda x:x.year*100 + x.month)
    benchmark = benchmark.groupby('ym').last()
    benchmark = benchmark.set_index('tradedate')[['price']]
    benchmark = benchmark.reindex(result.index)    
    benchmark['nav'] = benchmark['price']/benchmark.price[0]    


    
    result['benchmark'] = benchmark.nav.values
    result['RS'] = result.Portfolio/result.benchmark
    
    plt.figure(figsize = (8,4))
    plt.plot(result.Portfolio,linewidth = 2,c = 'deepskyblue',label = 'Portfolio')
    plt.plot(result.benchmark,linewidth = 2,c = 'darkred',label = 'benchmark')    
    plt.plot(result.RS,linewidth = 2,c = 'orange',label = 'Portfolio/benchmark')    
    plt.legend(fontsize = 12)

    return stockret.reset_index(drop = True),result
```
