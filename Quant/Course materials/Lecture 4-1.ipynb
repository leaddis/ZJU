Kernel具有如下操作：

Interrupt:  Causes the kernel to stop performing the current task without actually shutting the kernel down. You can use this option when you want to do something like stop processing a large dataset.

Restart: Stops the kernel and starts it again. This option causes you to lose all the variable data. However, in some cases, this is precisely what you need to do when the environment has become dirty with old data.

Restart & Clear Output: Stops the kernel, starts it again, and clears all the existing cell outputs

Restart & Run All: Stops the kernel, starts it again, and then runs every cell starting from the top cell and ending with the last cell. When Notebook reaches the bottom, it selects the last cell but doesn’t insert a new one.

Reconnect: Recreates the connection to the kernel. In some cases, environmental or other issues could cause the application to lose its connection, so you use this option to reestablish the connection without loss of variable data.

Shutdown: Shuts the kernel down. You may perform this step in preparation for using a different kernel.

Change Kernel: Selects a different kernel from the list of kernels you have installed. For example, you may want to test an application using various Python versions to ensure that it runs on all of them.

Python中有两种关于复制的概念，深复制与浅复制。深复制与浅复制是两种不同的复制方法：`.copy()`是对列表进行深复制，而赋值语句`=`是对列表进行浅复制。浅复制是赋值链接，复制对象与被复制对象是相关联的；深复制是完全赋值，复制对象与被复制对象是完全不相关联的两个对象。

```python
list1 = ['601939','601318','601398']
list2 = ['601939','601318','601398']

#深复制
list3 = list1.copy()
#浅复制
list4 = list2

# pop()方法根据索引值删除相应元素并返回被删除的元素，如果索引值缺省，则默认为-1，删除列表末尾的元素。
list3.pop()
list4.pop()

print('list1: ',list1)
print('list2: ',list2)
print('list3: ',list3)
print('list4: ',list4)
```

使用copy方法对list1深复制后，list1不受list3的影响，使用赋值语句对list2进行浅复制后，list2会受到list4的影响。这是因为在深复制下，系统给list3开辟了独立的空间地址，然后将list1中每个元素单独复制到list3的空间里，所以list1和list3是两个相互独立的列表。而在浅复制下，系统只复制list2的地址给list4，因此list2和list4的内容存储的空间地址都指向相同的地址，改变list4的内容会间接改变list2的内容。

### 集合（set）

python的set和其他语言类似, 是一个**无序不重复**元素集, 基本功能包括关系测试和消除重复元素. 集合对象还支持union(联合), intersection(交), difference(差)和sysmmetric difference(对称差集)等数学运算.

set 支持 `x in set`, `len(set)`,和 `for x in set`。作为一个无序的集合，set不记录元素位置或者插入点。因此，set不支持 indexing, slicing, 或其它类序列（sequence-like）的操作。

集合用于包含一组无序的对象。要创建集合，可使用

- `set(iterable)`函数
- `{元素1，元素2，元素3，...}`：指定具体的集合元素，元素之间以逗号分割

```python
s = set([3,5,9,10])      #创建一个数值集合
print(s)
```

```python
tt = list("Hello")
print(tt)

t = set("Hello")       #创建一个唯一字符的集合
print(t)
```

```python
s = {3,5,9,10}
```

与列表和元组不同，集合是无序的，也无法通过数字进行索引。此外，集合中的元素不能重复。例如，如果检查前面代码中t集合的值，结果会是：

```python
print(type(s))
print(s)

print(type(t))
print(t)
```

集合支持一系列标准操作，包括并集、交集、差集和对称差集，例如：

```python
a = t | s          # t 和 s的并集
print("a: ",a)

b = t & s          # t 和 s的交集
print("b: ",b)

c = t - s          # 求差集（项在t中，但不在s中）
print("c: ",c)

d = t ^ s          # 对称差集（项在t或s中，但不会同时出现在二者中）
print("d: ",d)
```

```python
t.add('x')            # 添加一项
print("t: ",t)

s.update([10,37,42])  # 在s中添加多项
print("s: ",s)

t.remove('H')         #使用remove()可以删除一项
print("t: ", t)
```

### break语句

break语句是用来终止程序的执行的。当循环结构中出现break语句时，即使在循环中没有false条件，或者序列还没有没完全递归完，也会终止循环语句。如果break语句出现在嵌套循环中，则应当终止执行最深层次的循环，并开始执行下一行代码。

```python
## 打印1到20的偶数，遇到10的整数偶数时，结束整个循环
i = 1
while i <= 20:
    if i % 2 == 0:
        if i % 10 ==0:
            break
        print(i)
    i += 1
```

```python
## break 是用来跳出其所在的最深层次的循环，如在双层循环语句中，用来跳出最内层的循环
a = 5
while a <= 8:
    a = a + 1
    for i in range(1,4):
        print(a,i)
        if i == 2:
            break
```

### continue 语句

continue 语句的功能是强制停止循环中的这一次执行，直接跳到下一次执行。即跳过当前循环的剩余语句，继续进行下一轮循环。

```python
# 打印小于10的所有奇数
n = 0
while n < 10:
    n = n+1
    if n % 2 == 0: # 如果n是偶数，则执行continue语句
        continue   # continue语句会直接继续下一轮循环，后续的print()不会执行
    print(n)
# 输出结果将全部为奇数，因为n为偶数时跳出本次循环。
```

```python
# continue在for循环中的应用，以上结果可以同样采用for循环得到
for n in range(1,10):
    if n % 2 == 0:
        continue
    print(n)
```

```python
## continue在双层循环语句中的实例
a = 1
while a < 7:
    a = a + 1
    if a == 4:
        continue
    for i in range(7,10):
        if i == 9:
            continue
        print(i)
```

### continue 与 break的区别

continue指的是结束本次循环中剩余的语句，然后继续执行下一轮的循环，而break语句指的是直接结束这个循环，包括结束执行该循环的剩余所有次循环

```python
# code 1
for i in range(10,19):
    if i == 15:
        continue
    print(i)
```

```python
# code 2
for i in range(10,19):
    if i == 15:
        break
    print(i)
```

在程序2中，i=15及之后均没有输出，因为执行遇到break，直接跳过了剩余所有循环，而程序1
只跳过了i = 15这一次循环。

## 推导式（comprehensions）

推导式（comprehensions），是Python的一种独有特性。推导式可以从一个数据序列构建另一个新的数据序列的结构体。推导式有三种，分别是列表（list）推导式，字典（dic）推导式和集合（set）推导式。

### 列表（list）推导式

列表推导式提供了一种简明扼要的方法来创建列表。其结构是在一个中括号里包含一个表达式，然后是一个for语句，然后是0个或多个for或者if语句。表达式可以是任意的，意思是你可以在列表中放入任意类型的对象。返回结果将是一个新的列表，在这个以if和for语句为上下文的表达式运行完成之后产生。

```python
'''
基本语法结构：

result = [out_exp_res for out_exp in input_list if condition]

out_exp_res：列表生成元素表达式，可以是有返回值的函数。
for out_exp in input_list：迭代input_list将out_exp传入out_exp_res表达式中。
if condition：根据条件过滤哪些值可以。（注：列表推导式中的if condition条件可以省略）

以上表达式等价于：
result = []
   ...: for out_exp in input_list:
   ...:     if condition:
   ...:         result.append(out_exp_res)
   

'''

# 过滤掉长度小于等于3的字符串列表，并将剩下的转换成大写字母
names = ['Bob','Tom','alice','Jerry','Wendy','Smith'] 

# 使用for循环实现
names_new = []
for name in names:
    if len(name) > 3:
        names_new.append(name.upper())
print("result_for: ",names_new)

# 使用列表推导式实现
names_new_c = [name.upper() for name in names if len(name)>3]
print("result_com: ",names_new_c)
```

```python
# 创建一个包含收盘价格的list
price_array = ['30.14','29.58','26.36','32.56','32.82']

# 使用for循环创建对应收盘价格的日期
date_array = []
date_base = 20170118
for i in range(len(price_array)):
    date_array.append(str(date_base))
    date_base += 1
print(date_array)
```

```python
# 使用列表推导式，注意这里if condition可以省略
date_base = 20170118
date_array = [str(date_base+ind) for ind in range(len(price_array))]
print(date_array)
```

```python
# 构造一个新的列表，将收盘价价和日期放在一起
# zip命令同时迭代多个序列，每次分别从一个序列中取一个元素，一旦其中某个序列达到结尾，则迭代结束。
stock_list = [[date,price] for date, price in zip(date_array, price_array)]
print(stock_list)
```

这里我们构建了一个新的列表stock_list,它的每一个元素是也一个列表，包括了某一天的日期和收盘价。

### 字典(dic)推导式

字典推导和列表推导的使用类似，只不过中括号改成大括号。

```python
'''
基本格式：{ key_expr: value_expr for value in collection if condition }

''' 
```

```python
'''
enumerate() method adds a counter to an iterable object and returns it in a form of enumerate object. 
This enumerate object can then be used directly in for loops or be converted into a list of tuples 
using list() method.

Syntax:

enumerate(iterable, start=0)

Parameters:
Iterable: any object that supports iteration
Start: the index value from which the counter is 
       to be started, by default it is 0 
'''

strings = ['import','is','with','if','file','exception']

# 使用list命令将enumerate对象变成list
print(list(enumerate(strings)))

# 使用list命令将enumerate对象变成list，可以看出list的每个元素都是tuple
type(list(enumerate(strings))[1])
```

关于enumerate的print命令有两种方式，第一，直接打印tuple对象，第二，将tuple元素拆开打印

```python
# printing the tuples in object directly 
for cv in enumerate(strings): 
    print(cv) 

# changing index
for cv in enumerate(strings,100): 
    print(cv)
    
# printing separately 
for count,value in enumerate(strings): 
    print(count, value) 
```

```python
# 通过字典推导式创建字典
strings = ['import','is','with','if','file','exception']
D = {key: val for key,val in enumerate(strings)}
print(D)
```

```python
# 构造一个新的字典，将收盘价价和日期放在一起
stock_dic = {date:price for date, price in zip(date_array, price_array)}
print(stock_dic)
```

## 参数

参数是函数在执行功能时所用到的数据。参数分为两种，一种是实际参数，一种是形式参数。形式参数一般发生在函数定义的过程中，一般是指参数的名称，不代表函数的值，只是形式上的参数，只表示一个函数中的哪个位置有哪个名称的参数。实际参数一般在调用的过程中，一般指具体的参数值，当调用函数时，形实际参数的值会被传递给形式参数，之后函数内部才能对形式参数进行存取，函数才能执行。

```python
def function2(a,b):
    if a > b:
        print(a)
    else:
        print(b)
        
function2(4,5)
#print(result_ab)
```

这里函数的名称是function2，其中有a,b两个形式参数，4和5是实际参数，分别赋值给了a和b。表达式也可以作为实际参数传递进函数中：

```python
def my_abs(x):
    if x >= 0:
        return x
    else:
        return -x

a = -2

result_a = my_abs(a)

print(result_a)

#abs(1 - a)
```

参数的传递方式有两种，一种是按顺序传递：

```python
def function3(a=1,b=2):
    if a > b:
        print(a)
    else:
        print(b)
        
function3(4,5)
```

这里a = 1,b =2定义了a和b的默认值，也即是说如果我们不写出参数，参数将使用默认值:

```python
function3()
```

另外一种是按照参数的名字传递参数，这就是关键字赋值，就是在调用时就写明实际参数和形式参数的绑定。关键字赋值不受赋值顺序影响

```python
function3(a=4, b=5)
function3(b=5, a=4)
```

## 可选参数 --- 参数个数未定

当函数的参数数目并不确定时，可以通过list，tuple，dict等进行传递

```python
def my_addition0(addend):
    sum = 0
    for i in addend:
        sum = sum + i
    print(sum)

numbers0 = (1,2,3)

numbers1 = [1,2,3]

my_addition0(numbers0)

my_addition0(numbers1)
```

这种方式的缺点在于要把传入的参数打包才能传递，否则Python会报错

```python
#my_addition(1,2)
```

如果不想事先打包传递的参数，可以在不确定数目的形式参数前加一个星号"\*"或者两个星号"\*\*"。两者的差异在于一个星号的形参为空的tuple或者list，而两个星号的形参为空的dict。

```python
def my_addition1(*addend):
    sum = 0
    for i in addend:
        sum = sum + i
    print(sum)

my_addition1(1,2)

my_addition1(1,2,3)
```

使用不定数目参数的另外一种方式是，函数前面N个参数是固定的，之后的参数数目则不一定，即f(x1,x2,...,xN, \*y)。例如要计算一个加权和，前面两个参数的权重分别为0.4和0.3，后面参数的个数为0.3除以后面参数的个数：

```python
def weighted_sum(x1,x2,*y):
    sum = 0
    size = len(y)
    weight = 0.3/size
    for i in y:
        sum = sum + weight*i
    sum = sum + 0.4*x1 + 0.3*x2
    return sum

weighted_sum(6,7,8,9)
```

## 作用域

在Python中，变量是在一定范围内起作用的，这个范围叫做作用域，如果一个变量名在代码文件内部、函数外部被赋值和创建，则该变量的作用域是全局的，在文件内部的任何地方都起作用：

```python
i = 10            # 定义一个全局变量i
print (i)

def function4():
    j = i + 1
    print(j)         

function4()

print(i)
```

可以看出在function4函数中对i的取值仍然是10，也就是全局变量的值，全局变量在函数内部依然起作用。需要注意的是在函数内部，虽然可以访问全局变量，但是不能对全局变量进行修改：

```python
i = 10            # 定义一个全局变量i
print(i)

def function4_1():
    i = i + 1
    print(i)         

function4_1()

#print(i)
```

作用域在一定范围内（非全局内）起作用的变量叫做局部变量。若变量名在函数内部被赋值与创建，则该变量的作用域是局部的，只在函数内部被访问，在函数外部不起作用。在一个函数中，若没有对变量进行全局变量声明，则默认为一个局部变量。

```python
i = 2                #给全局变量i赋值

def function5():
    i = 7            #给局部变量i赋值
    print(i)

function5()          # 打印局部变量i

print(i)             # 打印全局便量i
```

可以看出，如果局部变量和全局变量重名，那么在文件内部，函数外部,取到的将是全局变量，但是在函数内部，则取局部变量。如果需要在函数中给全局变量赋值，就需要用global关键字声明

```python
def function6():
    global i
    i = 10
    
function6()
print(i)
```

做了全局变量声明后，就可以在函数内部对全局变量进行修改：

```python
i = 10
print (i)

def function6_1(): # 注意函数function6_1和函数function4_1的区别
    global i
    i = i + 1
    
function6_1()
print(i)
```

# 面向对象

Python中支持两种编程方式来写代码，分别是：函数式编程、面向对象式编程，其他很多语言只支持一种。

函数式编程

```python
# 定义函数，在函数中实现功能
def func():
    print("Hello World")

# 执行函数
func()
```

面向对象式编程

```python
# 定义类
class Foo(object):
    # 在类中定义方法
    def func(self):
        print("Hello World")
    
# 实例化类的对象
obj = Foo()
# 通过对象调用类中的实例方法
obj.func()
```

想要通过面向对象去实现某个或某些功能时需要2步：

- 定义类，在类中定义方法，在方法中去实现具体的功能。

- 实例化类成为一个对象，通过对象去调用并执行方法。

```python
class Message:

    def send_email(self, email, content):
        data = "给{}发邮件，内容是：{}".format(email,content)
        print(data)


msg_object = Message() # 实例化一个对象 msg_object，创建了一个一块区域。
msg_object.send_email("econ@econ.com","注册成功")
```

注意事项：1.类名称首字母大写；2.在类中编写的函数称为方法；3.每个方法的第一个参数是`self`。

类中可以定义多个方法

```python
# 定义类
class Message:

    def send_email(self, email, content):
        data = "给{}发邮件，内容是：{}".format(email, content)
        print(data)

    def send_wechat(self, vid, content):
        data = "给{}发微信，内容是：{}".format(vid, content)
        print(data)

# 类实例化成对象
msg_object = Message()

# 通过对象调用方法
msg_object.send_email("econ@econ.com", "注册成功")
msg_object.send_wechat("econ", "注册成功")
```

## 对象和self

在每个类中都可以定义个特殊的：`__init__ 初始化方法 `，在实例化类创建对象时自动执行，即：`对象=类()`。

```python
class Message:

    def __init__(self, content):
        self.data = content

    def send_email(self, email):
        data = "给{}发邮件，内容是：{}".format(email, self.data)
        print(data)

    def send_wechat(self, vid):
        data = "给{}发微信，内容是：{}".format(vid, self.data)
        print(data)

# 对象 = 类名() # 自动执行类中的 __init__ 方法。

# 1. 根据类型创建一个对象，分配内存的一块区域 。
# 2. 执行__init__方法，模块会将创建的那块区域的内存地址当self参数传递进去，往区域中写入(data="注册成功")

# 类实例化成对象，自动执行_init_方法
msg_object = Message("注册成功")

# 利用对象调用方法
msg_object.send_email("econ@econ.com") # 给econ@econ.com发邮件，内容是：注册成功
msg_object.send_wechat("econ") # 给econ发微信，内容是：注册成功
```

函数式的思想：函数内部需要的数据均通过参数的形式传递。

面向对象的思想：将一些数据封装到对象中，在执行方法时，再去对象中获取。
- self，本质上就是一个参数。这个参数是Python内部会提供，其实本质上就是调用当前方法的那个对象。
- 对象，基于类实例化出来”一块内存“，默认里面没有数据；经过类的 `__init__`方法，可以在内存中初始化一些数据。

## 成员

在编写面向对象相关代码时，最常见成员有：

- 实例变量，属于对象，只能通过对象调用。
- 绑定方法，属于类，通过对象调用或通过类调用。

```python
class Person:

    def __init__(self, n1, n2):
        # 实例变量
        self.name = n1
        self.age = n2

    # 绑定方法
    def show(self):
        msg = "我叫{}，今年{}岁。".format(self.name, self.age)
        print(msg)

    def all_message(self):
        msg = "我是{}人，我叫{}，今年{}岁。".format(Person.country, self.name, self.age)
        print(msg)

    def total_message(self):
        msg = "我是{}人，我叫{}，今年{}岁。".format(self.country, self.name, self.age)
        print(msg)
```

```python
# 执行绑定方法
# 类实例化成对象
p1 = Person("John",20)
# 对象调用方法
p1.show()
```
